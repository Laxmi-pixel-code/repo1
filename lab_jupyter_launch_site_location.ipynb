```python
# Task 2: Mark the success/failed launches for each site on the map
# Create a new map object centered near the Cape Canaveral sites for a good initial view
# Re-initializing the map to remove any previous test markers (like the NASA JSC one)
nasa_coordinate = [29.5596688840303, -95.083879197059]
site_map = folium.Map(location=nasa_coordinate, zoom_start=10)

# The data for this task must come from the main DataFrame: spacex_df
# The 'class' column is crucial: 1=Success, 0=Failure

for index, row in spacex_df.iterrows():
    # Get the coordinates and launch outcome
    latitude = row['Lat']
    longitude = row['Long']
    launch_class = row['class']
    
    # 2. Determine the marker color based on the launch outcome ('class')
    # Success (class=1) -> Green
    # Failure (class=0) -> Red
    if launch_class == 1:
        color = 'green'
    else:
        color = 'red'
        
    # 3. Create a Folium CircleMarker for the launch attempt
    # We use CircleMarker because many launches occur at the exact same Lat/Long, 
    # and CircleMarkers are better for visually representing many points.
    
    # Define a custom icon (optional, but good for differentiation)
    icon = folium.DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12; color:' + color + ';">' + str(launch_class) + '</div>',
    )
    
    # Create and add the Marker with the colored icon
    folium.Marker(
        [latitude, longitude],
        icon=icon,
        popup=f"Launch Site: {row['Launch Site']}, Outcome: {'Success' if launch_class == 1 else 'Failure'}"
    ).add_to(site_map)

# Display the map (The output should show red/green markers clustered at the launch sites)
site_map
```

```python
import folium
# Ensure folium is installed in your environment before running this cell
# e.g., by running: !pip install folium or await piplite.install('folium')
from folium.plugins import MarkerCluster
from math import sin, cos, sqrt, atan2, radians

# --- Vincenty Distance Function ---
def calculate_distance(lat1, lon1, lat2, lon2):
    """Calculates the distance between two points (lat, lon) in kilometers."""
    R = 6373.0  # approximate radius of earth in km
    lat1_rad = radians(lat1)
    lon1_rad = radians(lon1)
    lat2_rad = radians(lat2)
    lon2_rad = radians(lon2)
    dlon = lon2_rad - lon1_rad
    dlat = lat2_rad - lat1_rad
    a = sin(dlat / 2)**2 + cos(lat1_rad) * cos(lat2_rad) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance = R * c
    return distance
# -----------------------------------------------------------------------------

# 1. Define the Launch Site Coordinate (CCAFS SLC-40)
launch_site_lat = 28.563197
launch_site_lon = -80.576820

# 2. Define the Target Feature Coordinate (Nearest Highway - US-1)
highway_lat = 28.57353
highway_lon = -80.65545

# 3. Calculate the distance
distance_highway = calculate_distance(launch_site_lat, launch_site_lon, highway_lat, highway_lon)

print(f"Distance from CCAFS SLC-40 to nearest Highway: {distance_highway:.2f} km")


# --- Part 2: Visualization on Map ---

# Re-initialize the map centered on the launch site for better view (high zoom)
site_coord = [launch_site_lat, launch_site_lon]
distance_map = folium.Map(location=site_coord, zoom_start=12)

# 4. Create a Marker for the Highway Point
highway_marker = folium.Marker(
    [highway_lat, highway_lon],
    icon=folium.Icon(color='orange', icon='road', prefix='fa'),
    popup=f"Nearest Highway: {distance_highway:.2f} km"
).add_to(distance_map)

# 5. Create a CircleMarker for the Launch Site
folium.Circle(
    site_coord,
    radius=100,
    color='red',
    fill=True,
    fill_color='#d34358',
    popup='CCAFS SLC-40 Launch Site'
).add_to(distance_map)

# 6. Draw a PolyLine to show the shortest distance
coordinates = [
    [launch_site_lat, launch_site_lon],
    [highway_lat, highway_lon]
]
folium.PolyLine(
    coordinates, 
    color='orange', 
    weight=5, 
    tooltip=f"Distance to Highway: {distance_highway:.2f} km"
).add_to(distance_map)

# Display the map
distance_map
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```
