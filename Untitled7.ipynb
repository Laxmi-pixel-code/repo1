```python
import dash
from dash import html
from dash import dcc
from dash.dependencies import Input, Output
import pandas as pd
import plotly.express as px

# -----------------
# 1. Data Setup and Parameters
# -----------------
# NOTE: Replace 'sample_data.csv' with the actual file name from your lab.
try:
    df = pd.read_csvdata = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv')
    print("Data loaded successfully.")
    
except FileNotFoundError:
    print("FATAL ERROR: Data file not found. Please verify the filename.")
    df = pd.DataFrame(columns=['Year', 'Vehicle_Type', 'Automobile_Sales', 'Advertising_Expenditure', 'unemployment_rate', 'Recession', 'Month'])

# Define the list of years (1980 to 2023)
year_list = [i for i in range(1980, 2024, 1)] 


# -----------------
# 2. Initialize the Dash app
# -----------------
app = dash.Dash(__name__)

# -----------------
# 3. Define the application Layout
# -----------------
app.layout = html.Div(
    [
        # Task 2.1: Main Title
        html.H1(
            "Automobile Sales Statistics Dashboard",
            style={'textAlign': 'center', 'color': '#503036', 'fontSize': 24}
        ),

        # Task 2.2: Select Report-type Dropdown
        dcc.Dropdown(
            id='dropdown-statistics',
            options=[
                {'label': 'Yearly Statistics', 'value': 'Yearly Statistics'},
                {'label': 'Recession Period Statistics', 'value': 'Recession Period Statistics'}
            ],
            placeholder='Select a report type',
            value='Select Statistics',
            style={'width': '80%', 'padding': '3px', 'fontSize': '20px', 'textAlignLast': 'center'}
        ),
        
        # Task 2.2: Select Year Dropdown
        dcc.Dropdown(
            id='select-year',
            options=[{'label': str(i), 'value': i} for i in year_list],
            placeholder='Select-year',
            value='Select-year',
            disabled=True 
        ),

        # Task 2.3: Output Division for Charts/Content
        html.Div(
            [], 
            id='output-container',
            className='chart-grid',
            style={'display': 'flex', 'flex-direction': 'column'}
        )
    ]
)

# -----------------
# 4. Callback Functions (Tasks 2.4, 2.5, 2.6)
# -----------------

# Callback 1: Enables/Disables the Select Year Dropdown
@app.callback(
    Output(component_id='select-year', component_property='disabled'),
    Input(component_id='dropdown-statistics', component_property='value')
)
def update_input_container(selected_statistics):
    if selected_statistics == 'Yearly Statistics':
        return False
    else:
        return True

# Callback 2: Generates the content for the output container
@app.callback(
    Output(component_id='output-container', component_property='children'),
    [
        Input(component_id='dropdown-statistics', component_property='value'),
        Input(component_id='select-year', component_property='value')
    ]
)
def update_output_container(selected_statistics, input_year):
    
    # Logic for Recession Period Statistics (Task 2.5)
    if selected_statistics == 'Recession Period Statistics':
        recession_data = df[df['Recession'] == 1]
        
        # Plot 1: Line Chart
        yearly_rec = recession_data.groupby('Year')['Automobile_Sales'].mean().reset_index()
        R_chart1 = dcc.Graph(
            figure=px.line(
                yearly_rec, x='Year', y='Automobile_Sales',
                title='Automobile Sales Fluctuation Over Recession Period (Yearly Mean)'
            )
        )

        # Plot 2: Bar Chart (Average Vehicles Sold)
        average_sales = recession_data.groupby('Vehicle_Type')['Automobile_Sales'].mean().reset_index()
        R_chart2 = dcc.Graph(
            figure=px.bar(
                average_sales, x='Vehicle_Type', y='Automobile_Sales',
                title='Average Number of Vehicles Sold by Vehicle Type During Recessions'
            )
        )

        # Plot 3: Pie Chart (Total Expenditure Share)
        exp_rec = recession_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum().reset_index()
        R_chart3 = dcc.Graph(
            figure=px.pie(
                exp_rec, values='Advertising_Expenditure', names='Vehicle_Type',
                title='Total Advertising Expenditure Share by Vehicle Type During Recessions'
            )
        )

        # Plot 4: Bar Chart (Effect of Unemployment Rate)
        unemp_data = recession_data.groupby(['unemployment_rate', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()
        R_chart4 = dcc.Graph(
            figure=px.bar(
                unemp_data,
                x='unemployment_rate', y='Automobile_Sales', color='Vehicle_Type',
                labels={'unemployment_rate': 'Unemployment Rate', 'Automobile_Sales': 'Average Automobile Sales'},
                title='Effect of Unemployment Rate on Vehicle Type and Sales'
            )
        )
        
        return [
            html.Div([R_chart1, R_chart2], style={'display': 'flex'}),
            html.Div([R_chart3, R_chart4], style={'display': 'flex'})
        ]

    # Logic for Yearly Statistics (Task 2.6)
    elif (input_year is not None) and (input_year != 'Select-year') and (selected_statistics == 'Yearly Statistics'):
        yearly_data = df[df['Year'] == input_year]

        # Plot 1 (Complete period chart - as defined in the provided snippets)
        yas_data = df.groupby('Year')['Automobile_Sales'].mean().reset_index()
        Y_chart1 = dcc.Graph(
            figure=px.line(
                yas_data, x='Year', y='Automobile_Sales',
                title='Yearly Automobile Sales using line chart for the whole period'
            )
        )

        # Plot 2: Line Chart (Average Sales by Month in selected year)
        average_sales = yearly_data.groupby('Month')['Automobile_Sales'].mean().reset_index()
        Y_chart2 = dcc.Graph(
            figure=px.line(
                average_sales, x='Month', y='Automobile_Sales',
                title=f'Average Automobile Sales by Month in {input_year}'
            )
        )

        # Plot 3: Bar Chart (Average Vehicles Sold by Vehicle Type in selected year)
        avr_vdata = yearly_data.groupby('Vehicle_Type')['Automobile_Sales'].mean().reset_index()
        Y_chart3 = dcc.Graph(
            figure=px.bar(
                avr_vdata, x='Vehicle_Type', y='Automobile_Sales',
                title=f'Average Vehicles Sold by Vehicle Type in the year {input_year}'
            )
        )
        
        # Plot 4: Pie Chart (Total Advertising Expenditure for Each Vehicle in selected year)
        exp_data = yearly_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum().reset_index()
        Y_chart4 = dcc.Graph(
            figure=px.pie(
                exp_data, values='Advertising_Expenditure', names='Vehicle_Type',
                title=f'Total Advertisement Expenditure for Each Vehicle in {input_year}'
            )
        )
        
        return [
            html.H3(f"Yearly Statistics Report for the year {input_year}", style={'textAlign': 'center'}),
            html.Div([Y_chart1, Y_chart2], style={'display': 'flex'}),
            html.Div([Y_chart3, Y_chart4], style={'display': 'flex'})
        ]


    # Default case
    else:
        return html.Div([])


# -----------------
# 5. Run the app
# -----------------
if __name__ == '__main__':
    app.run(debug=True)
```
